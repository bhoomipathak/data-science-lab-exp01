```python
import pandas as pd

sales = pd.Series(
    [600, 450, 800, 300],
    index=['Pen', 'Notebook', 'Bag', 'Pencil']
)

print("Values:")
print(sales.values)

print("Index:")
print(sales.index)

print("Bag sales:")
print(sales["Bag"])

print("Sales above 500:")
print(sales[sales > 500])

print("Pen" in sales.index)

sales_dict = sales.to_dict()
print("Dictionary:")
print(sales_dict)
```
